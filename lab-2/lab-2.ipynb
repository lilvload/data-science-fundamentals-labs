```python
import pandas as pd
import numpy as np
```

```python
football = pd.read_csv('2021-2022 Football Team Stats.csv', delimiter=';', encoding='latin-1')
```

**Context**

This dataset contains 2021-2022 football team stats.
Only teams of Premier League, Ligue 1, Bundesliga, Serie A and La Liga are listed.

**Columns' description**

Rk : Rank  
Squad : Squad's name  
Country : Name of the country  
LgRk : Squad finish within the league  
MP : Matches played  
W : Wins  
D : Draws  
L : Losses  
GF : Goals for  
GA : Goals against  
GD : Goal difference  
Pts : Points  
Pts/G : Points per game  
xG : Expected goals  
xGA : Expected goals allowed  
xGD : Expected goals difference  
xGD/90 : Expected goals difference per 90 minutes  
Attendance : Attendance per game during this season, only for home matches  
Top Team Scorer : Top team scorer in league play  
Goalkeeper : Goalkeeper with the most minutes in league play

### Умовний вибір

**Проста умова**

```python
median_attendance = football["Attendance"].median()
print('Медіанне відвідування на домашньому стадіоні:', int(median_attendance))

football.loc[football["Attendance"] > median_attendance]
```

**Складна умова**

```python
football.loc[(football["Attendance"] > median_attendance) & (football["Country"] == 'ESP'), ['Squad', 'Country', 'Attendance', 'Top Team Scorer', 'Goalkeeper']]
```

**Умовні селектори**

```python
football.loc[(football["Attendance"] > median_attendance) & (football["Country"].isin(['ESP', 'ENG'])), ['Squad', 'Country', 'Attendance', 'Top Team Scorer', 'Goalkeeper']]
```

```python
football.loc[football['Attendance'].isnull()]
```

### Редагування даних

```python
football['New Top Team Scorer'] = football['Top Team Scorer'].str.split("- ").str[0]
football['Top Team Scorer Goals'] = football['Top Team Scorer'].str.split("- ").str[1]
```

```python
football.drop('Top Team Scorer', axis=1, inplace= True)
```

```python
football.rename(columns={'New Top Team Scorer':'Top Team Scorer'}, inplace=True)
```
