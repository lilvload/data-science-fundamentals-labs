### Застосування відображень

**Метод map ()**

```python
movies[movies['Director'] == 'Quentin Tarantino']['Meta Score'].map(lambda m: m + 5)
```

```python
movies[movies['Director'] == 'Quentin Tarantino']['Meta Score']
```

**Метод apply()**

```python
akira_mean_rating = movies[movies['Director'] == 'Akira Kurosawa']['IMDb Rating'].mean()
akira_mean_rating
```

```python
high_score_mean = movies[movies['Meta Score'] > 50]['Meta Score'].mean()
high_score_mean
```

```python
def change_ratings(row):
    row['IMDb Rating'] = row['IMDb Rating'] - akira_mean_rating
    row['Meta Score'] = row['Meta Score'] - high_score_mean
    return row

movies.apply(change_ratings, axis='columns').head(5)
```
